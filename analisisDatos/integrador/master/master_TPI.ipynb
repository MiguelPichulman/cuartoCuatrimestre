### 1.2 Relación entre dato, información y conocimiento
- Dato: es la materia prima del analisis. Enetenderemos a un dato como un valor, una caracteristica o hecho asociado al fenomenno que se esta estudiando. Los datos, por si solos, no siempre tiene un significado claro y muchas veces aparcecen como valores aislados.
- Informacion: surge cunado los datos son organizados, procesados e interpretados de manera que adquieren significado.
- Conocimiento: aparece cuando la informacion se interpreta dentro un contexto y permite comprender mejor una situacion, explicar un fenomeno o tomar una decision fundamentada. Es la comprension que permite actuar con mayor fundamento.

### 1.3 Ciclo de vida del Analisis de datos
Es una secuencaia iterativa e interactiva de etapas. Iterativa porque el resultado de una etapa puede obligar a volver a pasos anteriores para mejorar la calidad del concimiento extraido, y es interactiva porque el analista participa activamente de todo el proceso.
<p>Se distinguen 6 etapas:
<p>- 1. Definición del problema: Todo análisis comienza con un problema a resolver, el cual puede ser un inconveniente, una pregunta específica o una necesidad del negocio. Todo el trabajo empieza una vez que se sabe a dónde se quiere llegar.
<p>- 2. Recolección de datos: Consiste en identificar y obtener el material de trabajo desde diversas fuentes de datos. Estas fuentes pueden ser bases de datos (consultas SQL), Data Warehouses, archivos de acceso público en la web o fuentes de empresas externas.
<p>- 3. Limpieza y preparación: Es considerada una etapa crítica, ya que busca garantizar que los datos estén en un estado apto para ser estudiados; si los datos están "sucios", los resultados serán de mala calidad. Incluye actividades como la integración de datos, el tratamiento de valores faltantes (nulos), la corrección de inconsistencias y la detección de valores atípicos.
<p>- 4. Análisis Exploratorio de Datos (EDA): En este paso se exploran los datos para resumirlos estadísticamente y se aplican técnicas de visualización para observar posibles patrones. También sirve para identificar anomalías, detectar problemas de calidad provenientes del paso anterior y plantear hipótesis preliminares.
<p>- 5. Modelado y evaluación: Los datos ya preparados se procesan utilizando algoritmos matemáticos (Aprendizaje Automático o Machine Learning, ya sea supervisado o no supervisado) para encontrar tendencias, patrones ocultos o predecir variables. Luego, estos modelos deben ser validados y evaluados correctamente para asegurar su calidad y efectividad.
<p>- 6. Comunicación y despliegue: Una vez completado el análisis, se elabora un informe final, gráficos o dashboards para comunicar el conocimiento descubierto a los interesados, a menudo apoyándose en el storytelling. Además, si se generaron modelos predictivos, estos pueden comenzar a utilizarse sobre nuevos datos (despliegue), requiriendo un monitoreo a largo plazo para validar que los patrones sigan siendo
